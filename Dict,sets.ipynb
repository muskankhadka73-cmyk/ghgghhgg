### SETS

```python
empty_set = set()
print(empty_set)
print(type(empty_set))
```

```python
numbers = {1,2,3,4,5,6,2,1,4,}
numbers
```

```python
colors_list = ["red", "green", "blue", "red", "yellow", "blue"]
colors_set = set(colors_list)
print(colors_set)
```

```python
chars_set = set("hello world")
print(chars_set)
```

```python
dir(chars_set)
```

```python
set_a = {1,2,3,4,5,6,7}
set_b= {0,2,9,4,5,8,7}
```

```python
# Operator: |
# Method: .union()
union_set_op = set_a | set_b
union_set_meth = set_a.union(set_b)

print(f"Union (|): {union_set_op}")
print(f"Union (method): {union_set_meth}")
```

```python
# Operator: &
# Method: .intersection()
intersect_set_op = set_a & set_b
intersect_set_meth = set_a.intersection(set_b)
print(f"Intersection (&): {intersect_set_op}")
print(f"Intersection (method): {intersect_set_meth}")
```
