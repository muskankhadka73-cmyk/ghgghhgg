```python
import random 
a = random.randint(1,6)
print(a)
user_input = ""
count = 1
while a != user_input:
    user_input = int(input("Enter the number:"))
    if count < 3:  
        if user_input > a:
            print("high")
            count +=1
        elif user_input < a:
            print("low")
            count +=1
        else:
            print("that's correct")
    else:
        print("you lost")
        break
        
```
