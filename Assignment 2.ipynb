### Assignment 2 - "The 'Cyber-Id' Generator"

```python
username = input("Enter Your Name: ")
birth_year = int(input("Enter your birth year: "))
is_member = ""
while is_member not in ("True", "False"):
    is_member = input("Are you a member, True or False: ")
if is_member == "True":
    print("You're a member.")
else: 
    print("You're not a member.")
```

```python
current_age = 2025 - birth_year
print(f"you're {current_age} years old.")
```

```python
entry_fee = 0
if is_member == "True":
    entry_fee +=10.5
    print(f" you're a member, your entry fee is: {entry_fee}")
else:
    entry_fee +=20.0
    print(f" you're not a member, your entry fee is: {entry_fee}")
```

```python
print(f"Username :{username} \nAge:{current_age} \nMember:{is_member} \nEntry Fee:${entry_fee}")
```
