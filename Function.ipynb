```python
# *args and **kwargs
# Use *args to pass multiple positional arguments.


def show_args(*args):   
    lists =""
    for item in args:
        print(item)
        lists += item
    return lists

value =show_args("111", "222")
print(value)
```

```python
def show_args(*args):   
    lists =0
    for item in args:
        lists += item
    return lists

value =show_args(1, 2, 3)
print(value)
```

```python
sum1 = 0
def addition(*args):
    for item in args:
        global sum1
        sum1 = sum1 + item
    return sum1

addition(1,2,4,22,44,55,55)
```

```python
# Use **kwargs to pass multiple keyword arguments.

name = input("Enter the Name:")
def show_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

show_kwargs(name=name, age=22, surname= "Doe", grade="10")
show_kwargs(name=name, age=24, surname= "Sharma", grade="11")
```

```python
def show_kwargs(*args,**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

    for item in args:
        print(item)

show_kwargs(1,2,3,4,5,name="Alice", age=22, surname= "Doe", grade="10")
```

```python
def show_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} for {value}")

show_kwargs(a="Apple", b="ball", c= "Cat", d="dog", e= "elephant")
```
