```python
import random 
a = random.randint(1,6)
print(a)
user_input = ""
count = 0
while a != user_input:
    user_input = int(input("Enter the number:"))
    count +=1
    if user_input > a:
        print("high")
    elif user_input < a:
        print("low")
    else:
        print("that's correct")

print("you got it.")
```
